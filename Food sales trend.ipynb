## Data cleaning

### Subtask:
Clean the data by handling missing values, removing duplicates, and correcting inconsistencies.


**Reasoning**:
Handle missing values, remove duplicates, and correct inconsistencies in the data.



```python
# Handle Missing Values
print("Missing values before handling:")
print(df.isnull().sum())

numerical_cols = ['Price', 'Rating', 'Delivery Time(Min)', 'Quantity', 'Discount Applied']
categorical_cols = ['Item Name', 'Platform', 'Location', 'Delivery Type', 'Payment Mode', 'Restaurant Name', 'Food Category', 'Day of Week']

for col in numerical_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].median(), inplace=True)

for col in categorical_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].mode()[0], inplace=True)

print("\nMissing values after handling:")
print(df.isnull().sum())


# Remove Duplicates
print(f"\nNumber of rows before removing duplicates: {len(df)}")
df.drop_duplicates(inplace=True)
print(f"Number of rows after removing duplicates: {len(df)}")


# Correct Inconsistencies
print("\nUnique values in 'Platform' before standardization:")
print(df['Platform'].unique())
df['Platform'] = df['Platform'].str.lower()
print("\nUnique values in 'Platform' after standardization:")
print(df['Platform'].unique())

print("\nUnique values in 'Item Name' before standardization:")
print(df['Item Name'].unique())
df['Item Name'] = df['Item Name'].str.lower()
print("\nUnique values in 'Item Name' after standardization:")
print(df['Item Name'].unique())

# Convert 'Order Date & Time' to datetime objects
df['Order Date & Time'] = pd.to_datetime(df['Order Date & Time'], errors='coerce')
df.dropna(subset=['Order Date & Time'], inplace=True)
```

## Data wrangling

### Subtask:
Create new features and group the data for further analysis.


**Reasoning**:
Create new features, group the data, and calculate the total quantity of each food item sold on each platform and each day.



```python
# Feature Engineering
df['Order Date & Time'] = pd.to_datetime(df['Order Date & Time'])
df['day_of_week'] = df['Order Date & Time'].dt.day_name()
df['month'] = df['Order Date & Time'].dt.month
df['time_of_day'] = pd.cut(df['Order Date & Time'].dt.hour, bins=[0, 6, 12, 18, 24], labels=['night', 'morning', 'afternoon', 'evening'], right=False)
df['is_weekend'] = df['Day of Week'].isin(['Saturday', 'Sunday'])

# Data Grouping
df_grouped = df.groupby(['Item Name', 'Platform', 'day_of_week'])['Quantity'].sum().reset_index()
display(df_grouped.head())
```

## Data analysis

### Subtask:
Analyze the grouped data to identify sales trends and insights.


**Reasoning**:
I need to perform the analysis requested in the subtask, including finding the top-selling items overall and by platform, comparing weekday and weekend sales, and performing statistical analysis.



```python
import pandas as pd
from scipy import stats

# 1. Overall Top-Selling Items
top_10_items = df_grouped.groupby('Item Name')['Quantity'].sum().sort_values(ascending=False).head(10)
print("Top 10 Overall Sold Items:\n", top_10_items)

# 2. Platform-Specific Top-Selling Items
for platform in df_grouped['Platform'].unique():
    top_5_platform = df_grouped[df_grouped['Platform'] == platform].groupby('Item Name')['Quantity'].sum().sort_values(ascending=False).head(5)
    print(f"\nTop 5 Items on {platform}:\n", top_5_platform)

# 3 & 4. Weekday vs. Weekend Sales Comparison and Top 20 Items
top_20_items = df_grouped.groupby('Item Name')['Quantity'].sum().sort_values(ascending=False).head(20).index
weekday_sales = {}
weekend_sales = {}

for item in top_20_items:
    item_data = df_grouped[df_grouped['Item Name'] == item]
    weekday_sales[item] = item_data[~item_data['day_of_week'].isin(['Saturday', 'Sunday'])]['Quantity'].mean()
    weekend_sales[item] = item_data[item_data['day_of_week'].isin(['Saturday', 'Sunday'])]['Quantity'].mean()

t_test_results = {}
for item in top_20_items:
    try:
      t_statistic, p_value = stats.ttest_ind(df_grouped[(df_grouped['Item Name'] == item) & (~df_grouped['day_of_week'].isin(['Saturday', 'Sunday']))]['Quantity'],
                                          df_grouped[(df_grouped['Item Name'] == item) & (df_grouped['day_of_week'].isin(['Saturday', 'Sunday']))]['Quantity'])
      t_test_results[item] = (t_statistic, p_value)
    except Exception as e:
        print(f"Error performing t-test for {item}: {e}")
        t_test_results[item] = (None, None)

print("\nWeekday vs Weekend Sales (Top 20 Items):")
for item, (t_statistic, p_value) in t_test_results.items():
  print(f"{item}: t-statistic={t_statistic:.2f}, p-value={p_value:.3f}")


# 5. Statistical Analysis
overall_avg_sales = df_grouped.groupby('Item Name')['Quantity'].mean()
overall_std_sales = df_grouped.groupby('Item Name')['Quantity'].std()

print("\nOverall Average and Standard Deviation of Sales:")
print(pd.DataFrame({'Average Sales':overall_avg_sales, 'Standard Deviation':overall_std_sales}))


# 6. Aggregate Data (Summary)
summary = pd.DataFrame({
    'Item': top_20_items,
    'Weekday Avg Sales': [weekday_sales[item] for item in top_20_items],
    'Weekend Avg Sales': [weekend_sales[item] for item in top_20_items],
    'Overall Avg Sales': overall_avg_sales,
    'Overall Std Sales': overall_std_sales
})

print("\nSummary of Top 20 Items:")
display(summary)
```

## Data visualization

### Subtask:
Create visualizations to represent the key findings from the data analysis.


**Reasoning**:
Generate visualizations based on the analysis results.  Create bar charts for overall sales trend, platform-specific sales, weekday vs. weekend sales, and statistical significance. Save each plot as an image file.



```python
import matplotlib.pyplot as plt

# 1. Overall Sales Trend
plt.figure(figsize=(12, 6))
top_10_items.plot(kind='bar', color='skyblue')
plt.title('Top 10 Overall Best-Selling Food Items')
plt.xlabel('Item Name')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('overall_sales_trend.png')
plt.show()

# 2. Platform-Specific Sales
plt.figure(figsize=(15, 10))
for i, platform in enumerate(df_grouped['Platform'].unique()):
    plt.subplot(1, 3, i + 1)
    top_5_platform = df_grouped[df_grouped['Platform'] == platform].groupby('Item Name')['Quantity'].sum().sort_values(ascending=False).head(5)
    top_5_platform.plot(kind='bar', color='lightcoral')
    plt.title(f'Top 5 Items on {platform}')
    plt.xlabel('Item Name')
    plt.ylabel('Total Quantity Sold')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
plt.savefig('platform_specific_sales.png')
plt.show()

# 3 & 4. Weekday vs. Weekend Sales and Top 20 Items Visualization
plt.figure(figsize=(16, 8))
plt.bar(top_20_items, [weekday_sales[item] for item in top_20_items], color='lightgreen', label='Weekday', yerr=[df_grouped[df_grouped['Item Name'] == item]['Quantity'].std() for item in top_20_items])
plt.bar(top_20_items, [weekend_sales[item] for item in top_20_items], bottom=[weekday_sales[item] for item in top_20_items], color='lightblue', label='Weekend', yerr=[df_grouped[df_grouped['Item Name'] == item]['Quantity'].std() for item in top_20_items])
plt.title("Average Sales of Top 20 Items: Weekday vs Weekend")
plt.xlabel('Food Item')
plt.ylabel('Average Quantity Sold')
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.savefig('weekday_weekend_sales.png')
plt.show()

# 5. Statistical Significance Visualization
plt.figure(figsize=(16, 6))
p_values = [t_test_results[item][1] for item in top_20_items if t_test_results[item][1] is not None]
colors = ['red' if p < 0.05 else 'blue' for p in p_values]
plt.bar(top_20_items, p_values, color=colors)
plt.axhline(y=0.05, color='black', linestyle='--', label='Significance Level (0.05)')
plt.title('T-test P-values for Weekday vs Weekend Sales (Top 20 Items)')
plt.xlabel('Food Item')
plt.ylabel('P-value')
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.savefig('statistical_significance.png')
plt.show()
```
